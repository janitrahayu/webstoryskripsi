```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
import gc
import time
import warnings
import logging
import joblib

logging.getLogger("lightgbm").setLevel(logging.ERROR)
warnings.filterwarnings('ignore')

from sklearn.model_selection import train_test_split, RandomizedSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score, mean_absolute_percentage_error

# Import Model Standar
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
from catboost import CatBoostRegressor
from sklearn.ensemble import GradientBoostingRegressor, ExtraTreesRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.linear_model import Ridge, Lasso, ElasticNet

# Import cuML (Versi GPU untuk Kaggle)
try:
    from cuml.ensemble import RandomForestRegressor as cuRF
    from cuml.svm import SVR as cuSVR
    CUML_AKTIF = True
    print("✅ RAPIDS cuML (GPU) berhasil dimuat untuk RF & SVR!")
except ImportError:
    from sklearn.ensemble import RandomForestRegressor as cuRF
    from sklearn.svm import SVR as cuSVR
    CUML_AKTIF = False
    print("⚠️ RAPIDS cuML tidak ditemukan. Beralih ke CPU standar untuk RF & SVR.")


# =====================================================================
# 1. PENGATURAN MODEL & HYPERPARAMETER TUNING
# =====================================================================
MODEL_PILIHAN = 'LightGBM'  # <--- GANTI MODEL DI SINI

daftar_model = {
    'XGBoost': (
        XGBRegressor(random_state=42, n_jobs=-1, tree_method='hist'),  # <--- Diperbarui dari 'gpu_hist' ke 'hist'
        {'learning_rate': [0.01, 0.05, 0.1], 'max_depth': [3, 5, 7], 'n_estimators': [100, 200]}
    ),
    'LightGBM': (
        LGBMRegressor(random_state=42, n_jobs=-1, verbose=-1, device='gpu'),
        {'learning_rate': [0.01, 0.05, 0.1], 'max_depth': [3, 5, 7], 'num_leaves': [15, 31, 63], 'n_estimators': [100, 200]}
    ),
    'CatBoost': (
        CatBoostRegressor(random_state=42, task_type='GPU', verbose=0),
        {'learning_rate': [0.03, 0.05, 0.1], 'depth': [4, 6, 8], 'iterations': [100, 200]}
    ),
    'Random Forest': (
        cuRF(random_state=42) if CUML_AKTIF else cuRF(random_state=42, n_jobs=-1),
        {'max_depth': [5, 10, 15], 'n_estimators': [100, 200]}
    ),
    'SVR': (
        cuSVR() if CUML_AKTIF else cuSVR(),
        {'C': [0.1, 1, 10, 50], 'kernel': ['linear', 'rbf']}
    )
}

# =====================================================================
# 2. FUNGSI VISUALISASI
# =====================================================================
def visualisasi_scatter(y_true, y_pred, title_skenario, nama_target="Produktivitas"):
    plt.figure(figsize=(6, 6))
    plt.scatter(y_true, y_pred, color='black', s=20, alpha=0.6)
    
    min_val, max_val = min(y_true.min(), y_pred.min()), max(y_true.max(), y_pred.max())
    margin = (max_val - min_val) * 0.05
    ax_min, ax_max = min_val - margin, max_val + margin
    
    plt.plot([ax_min, ax_max], [ax_min, ax_max], color='#2ca02c', linestyle='-.', linewidth=1.2, zorder=0)
    
    if len(y_true) > 1:
        z = np.polyfit(y_true, y_pred, 1)
        p = np.poly1d(z)
        x_trend = np.linspace(ax_min, ax_max, 100) 
        plt.plot(x_trend, p(x_trend), color='black', linestyle='--', linewidth=2.5, zorder=1)

    satuan = 't·ha⁻¹' if 'Produktivitas' in nama_target else 'Ha'
    plt.xlabel(f'Reference {nama_target} ({satuan})', fontsize=12, color='gray')
    plt.ylabel(f'Estimated {nama_target} ({satuan})', fontsize=12, color='gray')
    plt.xlim(ax_min, ax_max); plt.ylim(ax_min, ax_max)
    plt.title(f'{MODEL_PILIHAN} Multi-Output - {nama_target}\n{title_skenario}', fontsize=12, color='gray')
    plt.tight_layout()
    plt.savefig(f'/kaggle/working/Scatter_{MODEL_PILIHAN}_{nama_target}_{title_skenario}.png', dpi=300)
    plt.close()

def visualisasi_perbandingan_produksi(total_prediksi, target_resmi, nama_skenario):
    fig, ax = plt.subplots(figsize=(6, 5))
    bars = ax.bar(['Prediksi Model', 'Data Resmi'], [total_prediksi, target_resmi], color=['#2ecc71', '#7f8c8d'])
    
    for b, v in zip(bars, [total_prediksi, target_resmi]):
        ax.text(b.get_x() + b.get_width() / 2, v + (max(total_prediksi, target_resmi)*0.02), f"{v:,.0f} Ton", 
                ha='center', va='bottom', fontsize=10, fontweight='bold')
        
    selisih_persen = (total_prediksi - target_resmi) / target_resmi * 100
    ax.set_ylabel('Total Produksi (Ton)')
    ax.set_title(f'Produksi 2025 - {nama_skenario}\nSelisih: {selisih_persen:+.2f}%', pad=15)
    plt.tight_layout()
    plt.savefig(f'/kaggle/working/Bar_{MODEL_PILIHAN}_{nama_skenario}.png', dpi=300)
    plt.close()

def visualisasi_semua_skenario_produksi(hasil_list, target_resmi):
    if not hasil_list: return
    df_plot = pd.DataFrame(hasil_list)
    fig, ax = plt.subplots(figsize=(10, 6))
    
    labels = df_plot['Skenario'].tolist() + ['Data Resmi']
    values = df_plot['Total_Produksi_Prediksi_Ton'].tolist() + [target_resmi]
    colors = ['#1f77b4'] * len(df_plot) + ['#7f7f7f']
    
    bars = ax.bar(labels, values, color=colors, edgecolor='black')
    for b, v in zip(bars, values):
        ax.text(b.get_x() + b.get_width()/2, v + max(values)*0.01, f"{v:,.0f}", ha='center', va='bottom', fontsize=9)
        
    ax.axhline(target_resmi, color='red', linestyle='--')
    ax.set_title(f'Perbandingan Total Produksi 2025 ({MODEL_PILIHAN})', pad=15)
    plt.xticks(rotation=15, ha='right')
    plt.tight_layout()
    plt.savefig(f'/kaggle/working/Rekap_Semua_{MODEL_PILIHAN}.png', dpi=300)
    plt.show()

# =====================================================================
# 3. FUNGSI LATIH MODEL
# =====================================================================
def latih_dan_evaluasi_target(X_train, y_train, X_test, y_test, nama_model):
    model_instance, param_grid = daftar_model.get(nama_model, (None, None))
    
    grid_search = RandomizedSearchCV(
        model_instance, param_grid, cv=3, scoring='neg_mean_absolute_error', 
        n_iter=5, n_jobs=-1, random_state=42
    )
    grid_search.fit(X_train, y_train)
    best_model = grid_search.best_estimator_

    y_pred = best_model.predict(X_test)
    y_pred = np.maximum(0, y_pred)

    r2 = r2_score(y_test, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    mae = mean_absolute_error(y_test, y_pred)
    mape = mean_absolute_percentage_error(y_test, y_pred) * 100

    return best_model, y_pred, (r2, rmse, mae, mape)

# =====================================================================
# 4. DATA PREPARATION
# =====================================================================
PATH_2224 = "/kaggle/input/datasets/jejehayu/dataset-regresi-padi/Dataset_Siap_Modeling_Padi_P3_Fixed.csv"
PATH_2025 = "/kaggle/input/datasets/jejehayu/dataset-regresi-2025/Dataset_Regresi_2025.csv"
TARGET_PRODUKSI_RESMI_2025_TON = 1_583_262.0

df_2224 = pd.read_csv(PATH_2224 if os.path.exists(PATH_2224) else "Dataset_Siap_Modeling_Padi_P3_Fixed.csv")
df_2025 = pd.read_csv(PATH_2025 if os.path.exists(PATH_2025) else "Dataset_Regresi_2025.csv")

def bersihkan_df(df_temp):
    col_map = {c.lower(): c for c in df_temp.columns}
    
    if 'grid_id' in col_map and 'ID_Grid' not in df_temp.columns:
        df_temp.rename(columns={col_map['grid_id']: 'ID_Grid'}, inplace=True)
    if 'tahun_bulan' in col_map and 'Tahun_Bulan' not in df_temp.columns:
        df_temp.rename(columns={col_map['tahun_bulan']: 'Tahun_Bulan'}, inplace=True)

    if 'Tahun_Bulan' in df_temp.columns and 'Tahun' not in df_temp.columns:
        df_temp['Tahun'] = df_temp['Tahun_Bulan'].astype(str).str[:4].astype(int)
        df_temp['Bulan'] = df_temp['Tahun_Bulan'].astype(str).str[-2:].astype(int)
        
    return df_temp.sort_values(by=['ID_Grid', 'Tahun']).reset_index(drop=True)

df_2224 = bersihkan_df(df_2224)
df_2025 = bersihkan_df(df_2025)

fitur_s1 = [f for f in ['VV_median', 'VH_median', 'RVI_median', 'NDPI_median', 'CrossRatio_median'] if f in df_2224.columns]
fitur_s2 = [f for f in ['NDVI_median', 'EVI_median', 'LSWI_median', 'GNDVI_median', 'S2REP_median', 'BSI_median'] if f in df_2224.columns]
fitur_iklim = [f for f in ['Suhu_2m', 'Kelembapan_Tanah', 'Radiasi_Matahari', 'Curah_Hujan_P3', 'Curah_Hujan'] if f in df_2224.columns]

skenario_dict = {
    "S1_Saja": fitur_s1,
    "S2_Saja": fitur_s2,
    "Iklim_Saja": fitur_iklim,
    "S1_S2": fitur_s1 + fitur_s2,
    "Semua_Fitur": fitur_s1 + fitur_s2 + fitur_iklim
}

def ekstraksi_flatten(df_source, daftar_fitur, has_target=True):
    gData, coord, y_prod, y_luas, grid_ids = [], [], [], [], []
    expected_steps = df_source.groupby(['ID_Grid', 'Tahun']).size().max()

    for (id_grid, tahun), group in df_source.groupby(['ID_Grid', 'Tahun']):
        if len(group) == expected_steps:
            if has_target and 'Target_Produksi_Grid_Ton' in group.columns:
                tot_prod = group['Target_Produksi_Grid_Ton'].sum(skipna=True)
                luas = group['Luas_Padi_Ha'].max()
                if tot_prod > 0 and luas > 0:
                    gData.append(group[daftar_fitur].values.flatten())
                    coord.append(group[['Latitude', 'Longitude']].iloc[0].values)
                    y_prod.append(tot_prod / luas)
                    y_luas.append(luas)
                    grid_ids.append(id_grid)
            elif not has_target:
                gData.append(group[daftar_fitur].values.flatten())
                coord.append(group[['Latitude', 'Longitude']].iloc[0].values)
                grid_ids.append(id_grid)

    return np.array(gData), np.array(coord), np.array(y_prod), np.array(y_luas), grid_ids

# =====================================================================
# 5. LOOPING EKSEKUSI SKENARIO
# =====================================================================
hasil_evaluasi = []
hasil_produksi = []

print(f"🚀 MENJALANKAN EKSPERIMEN {MODEL_PILIHAN} DENGAN ALUR MULTI-OUTPUT")

for nama_skenario, daftar_fitur in skenario_dict.items():
    if len(daftar_fitur) == 0: continue
    print(f"\n{'='*75}")
    print(f"🔬 MENJALANKAN SKENARIO: {nama_skenario.upper()} ({len(daftar_fitur)} fitur dasar)")

    gc.collect()

    gData_2224, coord_2224, y_prod_2224, y_luas_2224, _ = ekstraksi_flatten(df_2224, daftar_fitur, True)
    gData_25, coord_25, _, _, grids_25 = ekstraksi_flatten(df_2025, daftar_fitur, False)

    if len(gData_2224) == 0: continue

    X_train_g, X_test_g, y_train_p, y_test_p = train_test_split(gData_2224, y_prod_2224, test_size=0.20, random_state=42)
    X_train_c, X_test_c, y_train_l, y_test_l = train_test_split(coord_2224, y_luas_2224, test_size=0.20, random_state=42)

    scaler_g = StandardScaler()
    X_train_g_sc = scaler_g.fit_transform(X_train_g)
    X_test_g_sc = scaler_g.transform(X_test_g)
    X_pred_g_sc = scaler_g.transform(gData_25)

    scaler_c = StandardScaler()
    X_train_c_sc = scaler_c.fit_transform(X_train_c)
    X_test_c_sc = scaler_c.transform(X_test_c)
    X_pred_c_sc = scaler_c.transform(coord_25)

    X_train_final = np.hstack([X_train_g_sc, X_train_c_sc]).astype(np.float32)
    X_test_final = np.hstack([X_test_g_sc, X_test_c_sc]).astype(np.float32)
    X_pred_final = np.hstack([X_pred_g_sc, X_pred_c_sc]).astype(np.float32)

    print(f"⏳ Melatih Model 1: Produktivitas ({MODEL_PILIHAN})...")
    model_prod, y_pred_p, metrik_p = latih_dan_evaluasi_target(
        X_train_final, y_train_p, X_test_final, y_test_p, MODEL_PILIHAN)

    print(f"⏳ Melatih Model 2: Luas Lahan ({MODEL_PILIHAN})...")
    model_luas, y_pred_l, metrik_l = latih_dan_evaluasi_target(
        X_train_final, y_train_l, X_test_final, y_test_l, MODEL_PILIHAN)

    hasil_evaluasi.append({
        'Skenario': nama_skenario,
        'R2_Prod': round(metrik_p[0], 4), 'RMSE_Prod': round(metrik_p[1], 4), 
        'MAE_Prod': round(metrik_p[2], 4), 'MAPE_Prod_%': round(metrik_p[3], 2),
        'R2_Luas': round(metrik_l[0], 4), 'RMSE_Luas': round(metrik_l[1], 4), 
        'MAE_Luas': round(metrik_l[2], 4), 'MAPE_Luas_%': round(metrik_l[3], 2)
    })

    print(f"✅ Evaluasi [Prod]: R2 = {metrik_p[0]:.4f} | RMSE = {metrik_p[1]:.4f} | MAE = {metrik_p[2]:.4f} | MAPE = {metrik_p[3]:.2f}%")
    print(f"✅ Evaluasi [Luas]: R2 = {metrik_l[0]:.4f} | RMSE = {metrik_l[1]:.4f} | MAE = {metrik_l[2]:.4f} | MAPE = {metrik_l[3]:.2f}%")
    
    visualisasi_scatter(y_test_p, y_pred_p, nama_skenario, "Produktivitas")
    visualisasi_scatter(y_test_l, y_pred_l, nama_skenario, "Luas_Lahan")

    if len(X_pred_final) > 0:
        print(f"🔮 Memproses prediksi 2025 untuk {len(X_pred_final)} grid...")
        
        y_pred_2025_prod = np.maximum(0, model_prod.predict(X_pred_final))
        y_pred_2025_luas = np.maximum(0, model_luas.predict(X_pred_final))

        df_hasil_pred = pd.DataFrame({
            'ID_Grid': grids_25,
            'Tahun': 2025,
            'Estimasi_Produktivitas_Ton_Ha': y_pred_2025_prod,
            'Estimasi_Luas_Padi_Ha': y_pred_2025_luas,
            'Estimasi_Produksi_Ton': y_pred_2025_prod * y_pred_2025_luas
        })

        total_produksi_ton = df_hasil_pred['Estimasi_Produksi_Ton'].sum()
        selisih_persen = (total_produksi_ton - TARGET_PRODUKSI_RESMI_2025_TON) / TARGET_PRODUKSI_RESMI_2025_TON * 100

        nama_file = f"/kaggle/working/Prediksi_2025_{MODEL_PILIHAN}_{nama_skenario}.csv"
        df_hasil_pred.to_csv(nama_file, index=False)
        print(f"📁 Data 2025 disimpan ke {nama_file}")

        hasil_produksi.append({
            'Skenario': nama_skenario,
            'Total_Produksi_Prediksi_Ton': round(total_produksi_ton, 2),
            'Target_Resmi_Ton': TARGET_PRODUKSI_RESMI_2025_TON,
            'Selisih_%': round(selisih_persen, 2)
        })

        visualisasi_perbandingan_produksi(total_produksi_ton, TARGET_PRODUKSI_RESMI_2025_TON, nama_skenario)

# =====================================================================
# 6. REKAPITULASI AKHIR
# =====================================================================
print("\n" + "="*85)
print(f"🏆 REKAPITULASI EVALUASI MULTI-OUTPUT ({MODEL_PILIHAN}):")
print("="*85)
if len(hasil_evaluasi) > 0:
    print(pd.DataFrame(hasil_evaluasi).to_string(index=False))

print("\n" + "="*85)
print("📊 REKAPITULASI TOTAL PRODUKSI (PRODUKTIVITAS x LUAS LAHAN DINAMIS) vs DATA RESMI:")
print("="*85)
if len(hasil_produksi) > 0:
    df_prod = pd.DataFrame(hasil_produksi).sort_values(by='Selisih_%', key=lambda s: s.abs())
    print(df_prod.to_string(index=False))
    
    visualisasi_semua_skenario_produksi(hasil_produksi, TARGET_PRODUKSI_RESMI_2025_TON)
```
